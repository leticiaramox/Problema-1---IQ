1. Nas primeiras duas imagens, enquanto não há colisões com a barreira de potencial, nota-se que não há perturbações na densidade de probabilidade $|\psi|^2$ e que sua propagação ocorre de forma organizada.
No terceiro gráfico, a função de onda torna-se deslocalizada ao entrar em contato com a barreira, indicando superposição de estados. A quarta imagem demonstra que, agora, há variação no formato da função de densidade com o tempo, fato que não ocorria anteriormente. Ademais, é visto perturbação à direita da barreira, demonstrando, assim, o fenômeno de tunelamento.

2. A presença de uma barreira de potencial altera qualitativamente o comportamento da função de onda, uma vez que modifica a equação de Schrödinger nas diferentes regiões do espaço. Em regiões onde o potencial é nulo, a função de onda assume a forma de ondas planas oscilatórias, representando as componentes incidente, refletida e transmitida. No interior da barreira, entretanto, o potencial 
V(x) impõe uma mudança no caráter das soluções. Para energias menores que a altura da barreira ($E \ll V_0$), a equação de Schrödinger admite soluções exponencialmente decrescentes, ou seja, ondas evanescentes, em vez de soluções oscilatórias. Essas soluções indicam que a amplitude da função de onda decai rapidamente dentro da barreira, mas nunca se anula completamente. Como a densidade de probabilidade é dada por $|\psi|^2$, segue-se que existe uma probabilidade finita de a partícula ser encontrada além da barreira, mesmo em uma região classicamente proibida.

3. Para uma onda vindo da esquerda
$$
\psi(x) =
\begin{cases}
A e^{ikx} + B e^{-ikx}, & \text{(região esquerda)} \\
C e^{iqx}, & \text{(região direita)}
\end{cases}
$$em que:

    A: amplitude de onda incidente;
    B: amplitude de onda refletida;
    C: amplitude de onda transmitida;
    k: número de onda antes da barreira;
    q: número de onda depois da barreira

    Sabe-se, também, por propriedades de reflexão e transmissão de ondas, que 
    $$
    R + T = 1
    $$
    Nesse caso, $T$ = coeficiente de transmissão e $R$ = coeficiente de reflexão. Como estamos focando apenas na parcela transmitida, admite-se a equação
    $$
    T = \frac{q}{k} \frac{|C|^2}{|A|^2}
    $$
    Logo, pelos gráficos obtidos:
    $$
    T \approx 0.04\%
    $$
    Perceba que, embora $T \ll R$, tem-se $T \neq 0$. Esse resultado é uma consequência direta da natureza ondulatória da partícula descrita pela equação de Schrödinger. Para energias menores que a altura da barreira, a função de onda no interior da barreira não se anula, mas decai exponencialmente, de modo que $|\psi|^2$ permanece diferente de zero. Portanto, existe uma probabilidade finita de a partícula ser encontrada do outro lado da barreira, caracterizando o fenômeno de tunelamento quântico.
    O princípio da incerteza de Heisenberg, $\Delta x \, \Delta p \geq \frac{\hbar}{2}$, fornece uma interpretação complementar: ao confinar a partícula em uma região do espaço, aumenta-se a incerteza no momento, permitindo componentes de momento suficientemente grandes para contribuir para a penetração na barreira.

4. No item 3, foi demonstrado que $T \neq 0$. Ou seja, fenômenos de transmissão são pequenos, mas nunca inexistentes.
   Esse é exatamente o fenômeno por trás do vazamento em transistores: as barreiras de potencial presentes neles, por exmplo, as junções PN, deveriam impedir a passagem de elétrons, contudo, através do tunelamento quântico, a função de onda não zera dentro da barreira, mas decai exponencialmente. Isso gera uma corrente pequena, mas finita, mesmo com o transistor desligado.

### 1. Trade-off dos Códigos Hamming
1.1. Tendo em mente que $R = k/n$ [1], para $(k, n) = (c, d)$, temos que para Hamming (7, 4):
$$
R_1 = 4/7 \approx 0.57
$$
Para Hamming (15, 11):
$$
R_2 = 11/15 \approx 0.73
$$
E para Hamming (31, 26):
$$
R_3 = 26/31 \approx 0.84
$$
Logo, o maior $R$ será $R_3$. Ademais, a análise do gráfico de desempenho efetivo em função da taxa de erro mostra que, embora códigos de Hamming mais longos apresentem maior taxa de informação, sua robustez em canais ruidosos é reduzida, uma vez que a probabilidade de ocorrência de múltiplos erros por bloco aumenta com o comprimento do código, limitando a eficácia da correção de erros. Dessa forma, o código de Hamming (7,4) apresenta maior efetividade em cenários com altas taxas de erro.

Já analisando a taxa SNR, Signal-to-Noise Ratio (em português, Relação Sinal-Ruído), códigos maiores tendem a ter SNR menores. Com isso, o código com a maior relação é o Hamming (7,4), o menor entre os três.

1.2. Em regimes de baixo SNR, a probabilidade de erro de bit ($p$) torna-se elevada. Como códigos com blocos longos agregam um maior número de bits em cada palavra-código, a probabilidade de ocorrência de dois ou mais erros em um mesmo bloco cresce significativamente. Como os códigos de Hamming corrigem apenas um erro por bloco, essa condição compromete sua eficácia, fazendo com que códigos mais curtos apresentem maior throughput efetivo em canais altamente ruidosos.

Essa afirmação pode ser corroborada pelo gráfico de Desempenho Efetivo versus Taxa de Erro: observa-se que, para $p = 0.5$, o código com maior throughput efetivo é o menor deles, o Hamming (7,4). Isso ocorre porque, para blocos menores, a probabilidade de múltiplos erros simultâneos é reduzida. Em contraste, para o Hamming (31,26), o desempenho efetivo é inferior nesse regime, evidenciando a maior vulnerabilidade de blocos longos a erros múltiplos.

Esse comportamento está diretamente relacionado ao trade-off entre capacidade de correção e overhead. Para valores elevados de $p$, torna-se necessário corrigir mais erros por bloco, o que exige o aumento da redundância do código (overhead). Entretanto, o aumento da redundância implica uma redução da taxa de transmissão útil, diminuindo o throughput efetivo. Essa relação é expressa por
$$
throughput = R\cdot(1 - P_{\text{falha}}),
$$
onde $R$ é a taxa do código. O aumento de $R$ implica a redução da redundância, enquanto a diminuição de $P_{\text{falha}}$ requer maior redundância. Assim, aumentar a redundância melhora a robustez do sistema, mas reduz a taxa efetiva de transmissão, enquanto reduzir a redundância aumenta a eficiência espectral, porém torna o sistema mais sensível ao ruído.

### 2. Regime de Operação

2.1. Para valores pequenos de $p$, o código de melhor desempenho é o Hamming (31,26). Isso ocorre porque, embora códigos maiores possuam mais bits de paridade em termos absolutos $(31−26>15−11>7−4)$, eles apresentam maior taxa de informação $R = k/n$, ou seja, menor redundância relativa. Em regimes de baixo ruído, a probabilidade de múltiplos erros por bloco é pequena, de modo que a maior eficiência espectral dos códigos longos resulta em um throughput efetivo superior.

2.2. Para valores maiores de $p$, os códigos mais eficientes são os menores, como o Hamming (7,4). Conforme discutido no item 1.2, em regimes de alta taxa de erro, a robustez do código torna-se mais importante do que a eficiência espectral. Como códigos de Hamming corrigem apenas um erro por bloco, a probabilidade de falha cresce rapidamente com o comprimento da palavra-código, pois blocos maiores acumulam mais bits e, portanto, maior probabilidade de múltiplos erros simultâneos.

Dessa forma, códigos mais curtos apresentam menor probabilidade de falha por bloco, o que resulta em maior throughput efetivo em canais altamente ruidosos.

### 4. Interpretação Energética

4.1. O throughput efetivo, no contexto dos códigos de Hamming, refere-se à taxa real de bits de informação úteis que são transmitidos com sucesso em um sistema de comunicação, levando em consideração tanto a sobrecarga introduzida pelos bits de paridade quanto as falhas de transmissão. Em outras palavras, ele representa a fração de dados efetivamente recuperados no receptor por unidade de tempo, descontando os bits adicionais utilizados para detecção e correção de erros. Assim, o throughput efetivo mede a eficiência prática do sistema, pois incorpora o trade-off entre redundância (overhead) e robustez contra erros, refletindo o quanto da capacidade do canal é, de fato, aproveitada para transmitir informação útil.

Além disso, o throughput efetivo constitui uma métrica adequada de desempenho, pois quantifica diretamente a eficiência operacional do esquema de codificação. Dessa forma, códigos com maior throughput efetivo apresentam maior eficiência na transmissão de dados, considerando simultaneamente confiabilidade e taxa de informação.

4.2. Ao reduzir a largura da barreira $a$, a corrente de vazamento por tunelamento quântico aumenta significativamente. Esse efeito eleva a taxa de erro de bits no circuito, exigindo maior redundância no código de correção de erros, conforme discutido em 3.2. Observa-se, no gráfico de Desempenho versus Tamanho do Transistor, que os valores de throughput efetivo são menores para valores reduzidos de $a$, evidenciando o impacto negativo do tunelamento quântico na eficiência da transmissão de dados.

### 5. Aplicação Prática

5.1. De acordo com o gráfico de Desempenho versus Tamanho do Transistor, em situações de transistores excessivamente pequenos, o código Hamming (7,4) apresenta maior throughput efetivo, dado que ele apresenta maior robustez em canais ruidosos.

5.2. De acordo com o gráfico de Desempenho versus Tamanho do Transistor, observa-se que, para valores maiores de $a$, códigos mais longos tornam-se mais eficientes. Nesse regime, a probabilidade de falha é reduzida, pois o tunelamento quântico e o ruído elétrico são menos significativos. Assim, códigos como o Hamming (31, 26), que apresentam maior taxa de informação $R$ e menor SNR, passam a ser mais vantajosos, pois conseguem transportar mais dados úteis sem comprometer a confiabilidade da transmissão.

5.3. Os códigos de correção não resolvem os problemas gerados pelo tunelamento, eles apenas os mitigam. O tunelamento aontece porque elétrons têm natureza ondulatória e podem atravessar barreiras finas mesmo sem energia clássica suficiente. Isso gera corrente de vazamento, ruído e erros físicos reais no hardware. Nenhum código digital pode impedir isso.

O que códigos como o Hamming fazem é detectar e corrigir os erros de bits causados pelo hardware e recuperar dados depois que o erro já aconteceu. Matematicamente, eles reduzem $P_{\text{falha}}$, mas não alteram $p$. 

Ademais, é fácil ver que existe um limite para a miniaturização de transistores: se o tunelamento aumenta demais, $p$ cresce muito, o que leva a blocos longos terem muitos erros simultâneos. Como Hamming só corrige 1 erro por bloco, $P_{\text{falha}$ explode e o throughput efetivo cai para quase 0. 

Atualmente, o limite prático é $a \approx 1-2 nm$. Abaixo disso, o vazamento por tunelamento explode, a correção de erro não salva e, com isso, o transistor deixa de ser transistor.

1.1. A probabilidade de tunelamento varia exponencialmente com a largura da barreira. É fácil ver que, conforme $a$ cresce, os elétrons enfrentarão uma maior dificuldade em atravessar o potencial imposto, diminuindo, assim, as correntes de vazamento.

1.2. Com a diminuição da largura da barreira, $a$, as correntes de vazamento no circuito implementado aumentam. Isso gera perda de informação e energia, o que eleva a taxa de erro de bit. Como consequência, a probabilidade de falha aumenta e o throughput efetivo do sistema é reduzido.

1.3. Como os transistores atualmente utilizados em circuitos integrados possuem dimensões extremamente reduzidas, sua disposição pode ser aproximada como bidimensional, isto é, cada transistor ocupa essencialmente uma área no plano do chip, em vez de um volume tridimensional. Dessa forma, a área ocupada por cada transistor escala como $a_{\text{transistor}} \propto a^2$, onde $a$ é a dimensão linear característica do dispositivo. Consequentemente, a densidade de transistores, definida como o número de transistores por unidade de área, é inversamente proporcional a essa área, resultando na relação aproximada $\rho \propto a^{-2}$. Mesmo que o layout real dos circuitos seja mais complexo, essa estimativa captura corretamente a ordem de grandeza do escalonamento geométrico.

1.4. Quando $a$ é muito pequeno, a corrente de vazamento associada ao tunelamento quântico torna-se significativa, resultando em perdas substanciais de energia e aumento da taxa de erro no sistema, como discutido anteriormente. Por outro lado, quando $a$ é muito grande, ocorre um desperdício de área no chip, reduzindo a densidade de transistores e tornando o sistema menos eficiente e pouco competitivo em termos de integração, o que dificulta a implementação de circuitos complexos e de alta performance.

Assim, existe um valor ótimo para a largura da barreira, que representa um compromisso entre eficiência elétrica (minimização do tunelamento e das perdas) e eficiência geométrica (maximização da densidade de transistores). Esse ponto ótimo define o regime de operação mais vantajoso para o projeto de semicondutores.

### 3. Trade-offs e Limites Físicos

3.1. Reduzir $a$ além do ponto ótimo implica em um aumento significatico da probabilidade de tunelamento e, consequentemente, da corrente de vazamento nos transistores. Como discutido anteriormente, isso resulta em perdas substanciais de energia e crescimento da taxa de erro no sistema.

3.2. O critério de eficiência máxima conduz ao menor valor de $a$. Isso ocorre porque ele desconsidera as falhas de transmissão e considera apenas a taxa do código $R$. Assim, a analogia adotada — ainda que idealizada e fisicamente irrealista — é que, quanto menor for $a$, maior será o número de transistores por área, resultando em maior densidade de integração e, portanto, maior eficiência aparente. Nesse critério, os efeitos do tunelamento quântico não são penalizados, o que leva a uma superestimação do desempenho do sistema em regimes de miniaturização extrema.

3.3. Como dito anteriormente, quando $a$ é muito pequeno, a corrente de vazamento associada ao tunelamento quântico torna-se significativa, resultando em perdas substanciais de energia e aumento da taxa de erro no sistema. Por outro lado, quando $a$ é muito grande, ocorre um desperdício de área no chip, reduzindo a densidade de transistores e tornando o sistema menos eficiente e pouco competitivo em termos de integração, o que dificulta a implementação de circuitos complexos e de alta performance.

### 4. Implicações Práticas

4.1. Para Iwai e Wong (2005)[2], o limite máximo da miniaturização de transistores é governado não pela tecnologia, mas pela física. Sabendo que a distância entre átomos individuais de silício (0,27 nm) é uma limitação intransponível, parâmetros do dispositivo, como corrente e tensão, também são limitados pelo princípio da incerteza. Uma variação de apenas uma camada atômica em um transistor com $a$ de 5 nm representaria uma mudança de aproximadamente 5,3% nesse comprimento, valor que pode exceder a margem de ruído aceitável para circuitos em escala giga (normalmente estimada em cerca de 5%). 

4.2. De acordo com Steane (1995)[3], códigos mais longos e redundantes permitem operar circuitos em regimes de maior taxa de erro, o que, em princípio, possibilita uma maior miniaturização dos transistores. Entretanto, essa miniaturização ocorre ao custo de maior overhead, menor taxa efetiva de transmissão e maior complexidade de hardware. Assim, códigos de correção de erro mitigam, mas não eliminam, o limite físico imposto pelo tunelamento quântico, estabelecendo um compromisso entre densidade de integração e eficiência de comunicação, assim como dito anteriormente.

4.3. Segundo Raisinghani (2001)[4], se chips fossem miniaturizados até a escala de dezenas de nanômetros, seu funcionamento seria prejudicado pelo surgimento de fenômenos quânticos, como o tunelamento de elétrons através das barreiras entre os fios. Ademais, Iwai e Wong (2005) afirmam que a separação entre a fonte e o dreno deve ser maior que o comprimento de onda de de Broglie dos elétrons, (aproximadamente 0,86 nm). Se essa condição não for satisfeita, os elétrons passam a se comportar mais como ondas do que como partículas, e o dispositivo deixa de operar como um transistor. 

Com isso, percebe-se que o limite máximo de miniaturização é imposto fisicamente nos componentes do circuito e não é uma questão de aprimoramento de software.

### 5. Extensão e Reflexão

5.1. Definiria eficiência considerando também os efeitos de tunelamento quântico. Como o objetivo atual da miniaturização de transistores é mitigar os efeitos quânticos indesejados, isso se torna particularmente relevante, visto que os componentes implementados em circuitos modernos operam em escalas da ordem de 1–3nm. Nessa faixa de dimensões, fenômenos quânticos, como o tunelamento eletrônico, afetam significativamente o processamento de informação e o consumo de energia, impactando diretamente a confiabilidade e o desempenho dos dispositivos.

5.2. Segundo Freitas et al. (2022)[5], a miniaturização dos dispositivos semicondutores aumenta a ocorrência de falhas múltiplas (MCUs), que não podem ser eficientemente corrigidas por códigos unidimensionais simples, como os códigos de Hamming. Por essa razão, códigos multidimensionais, em especial códigos produto bidimensionais, têm sido propostos, pois oferecem maior capacidade de detecção e correção de erros com custo proporcionalmente menor em termos de energia e área. Dessa forma, tem sido possível construir transistores com dimensões ainda menores do que aqueles cujo sistema de correção de erros se baseia apenas em códigos de Hamming.

5.3. Segundo Moore (1965), o número de transistores em um chip dobra aproximadamente a cada 18–24 meses, mantendo o custo por transistor praticamente constante. Entretanto, à medida que a tecnologia de fabricação alcançou a escala nanométrica, surgiram limitações fundamentais impostas pela física, que desafiam a continuidade dessa tendência.

Atualmente, os transistores comerciais operam com dimensões na ordem de 1–3 nm, onde efeitos quânticos tornam-se relevantes. Além disso, flutuações na posição de átomos individuais tornam-se comparáveis às dimensões do dispositivo, resultando em variações estatísticas nos parâmetros elétricos, aumento do ruído e redução da confiabilidade. Esses fenômenos indicam que, em escalas nanométricas, a física quântica passa a substituir a engenharia como principal fator limitante da miniaturização.

Assim, códigos de correção de erros (ECCs) mais sofisticados têm sido progressivamente implementados em circuitos integrados modernos, em resposta às limitações físicas impostas pela miniaturização dos dispositivos semicondutores.
